**Dataset:** Hair Type Classification (`curly=0`, `straight=1`, `800` train / `201` held-out evaluation images, SHA-256 `9e53453e3017502f22860cb08558f0cdb343e102fb61feab75960616185e7d67`)  
**Reference Environment:** Python `3.11`, `torch==2.9.0+cpu`, `torchvision==0.24.0+cpu`, `numpy==2.3.3`, `Pillow==11.3.0`, `SEED=42`

```python
import json
import statistics
from pathlib import Path
import numpy as np
from reference_train import HairModel

model = HairModel()
total_params = sum(p.numel() for p in model.parameters())
print("Q1 Loss Function: nn.BCEWithLogitsLoss()")
print(f"Q2 Total Trainable Parameters: {total_params:,}")
for name, param in model.named_parameters():
    print(f"  - {name}: {tuple(param.shape)} -> {param.numel():,}")
```

```python
baseline = json.loads(Path("history_baseline.json").read_text(encoding="utf-8"))
print("=== Baseline Training History (10 Epochs) ===")
for i in range(10):
    print(
        f"Epoch {i+1:02d} | train_loss={baseline['train_loss'][i]:.4f} | "
        f"train_acc={baseline['train_accuracy'][i]:.4f} | "
        f"eval_loss={baseline['evaluation_loss'][i]:.4f} | "
        f"eval_acc={baseline['evaluation_accuracy'][i]:.4f}"
    )

q3_median_acc = statistics.median(baseline["train_accuracy"])
q4_std_loss = float(np.std(baseline["train_loss"], ddof=0))
print(f"\nQ3 Median baseline train_accuracy: {q3_median_acc:.6f} -> {q3_median_acc:.2f}")
print(f"Q4 Population std of baseline train_loss (ddof=0): {q4_std_loss:.6f} -> {q4_std_loss:.3f}")
```

```python
augmented = json.loads(Path("history_augmented.json").read_text(encoding="utf-8"))
print("=== Augmented Training History (Epochs 1-10 of Augmented Phase) ===")
for i in range(10):
    print(
        f"Aug Epoch {i+1:02d} | train_loss={augmented['train_loss'][i]:.4f} | "
        f"train_acc={augmented['train_accuracy'][i]:.4f} | "
        f"eval_loss={augmented['evaluation_loss'][i]:.4f} | "
        f"eval_acc={augmented['evaluation_accuracy'][i]:.4f}"
    )

q5_mean_eval_loss = statistics.mean(augmented["evaluation_loss"])
q6_last5_eval_acc = statistics.mean(augmented["evaluation_accuracy"][-5:])
print(f"\nQ5 Mean augmented evaluation_loss: {q5_mean_eval_loss:.6f} -> {q5_mean_eval_loss:.3f}")
print(f"Q6 Mean last-5 augmented evaluation_accuracy (epochs 6-10): {q6_last5_eval_acc:.6f} -> {q6_last5_eval_acc:.2f}")
```
